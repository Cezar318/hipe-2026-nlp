````python
import os
import re
import json
from tqdm import tqdm
from google.colab import drive
from groq import Groq
from google.colab import userdata

# 1. Mount Google Drive safely
drive.mount('/content/drive')

DRIVE_REPO_PATH = '/content/drive/MyDrive/colab_data/HIPE-2026-data'
PATH_CLEANED_DEV = os.path.join(DRIVE_REPO_PATH, "data/sandbox/en-dev-cleaned.jsonl")

# 2. Add your Groq API key here (Get a free one instantly at console.groq.com)
GROQ_API_KEY = userdata.get("GROQ_API_KEY")
client = Groq(api_key=GROQ_API_KEY)

# Use Llama 3 8B (or change to "llama3-70b-8192" if you want the big model!)
MODEL_NAME = "llama-3.1-8b-instant"

def load_hipe_jsonl(path):
    data = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip(): data.append(json.loads(line))
    return data

def harvest_json_robust(text):
    if not text: return []
    clean_text = re.sub(r'[\x00-\x1F\x7F]', ' ', text)
    matches = re.findall(r'\{[^{}]*\}', clean_text)
    harvested_list = []
    for m in matches:
        try:
            m_clean = m.replace('TRUE', '"TRUE"').replace('FALSE', '"FALSE"').replace('PROBABLE', '"PROBABLE"')
            obj = json.loads(m_clean)
            if isinstance(obj, dict) and ("at" in obj or "isAt" in obj): harvested_list.append(obj)
        except: continue
    return harvested_list

print("📂 Loading dataset from Drive...")
cleaned_dev_docs = load_hipe_jsonl(PATH_CLEANED_DEV)
results = []

# Define your 4-shot examples string
examples_4_shot = """
EXAMPLE 1:
Text: "Captain Reynolds of Liverpool docked his schooner yesterday morning."
Pairs:
0. Person: Reynolds | Place: Liverpool
Results:
{"at": "TRUE", "isAt": "TRUE"}

EXAMPLE 2:
Text: "The local magistrate from Hamburg was traveling through the outskirts of Berlin."
Pairs:
0. Person: magistrate | Place: Hamburg
1. Person: magistrate | Place: Berlin
Results:
{"at": "TRUE", "isAt": "FALSE"}
{"at": "PROBABLE", "isAt": "TRUE"}

EXAMPLE 3:
Text: "Elizabeth Smith passed away at her country estate in Devonshire."
Pairs:
0. Person: Elizabeth Smith | Place: Devonshire
Results:
{"at": "TRUE", "isAt": "TRUE"}

EXAMPLE 4:
Text: "The French delegation, representing Paris, met with officials out in regional Versailles."
Pairs:
0. Person: French delegation | Place: Paris
1. Person: French delegation | Place: Versailles
Results:
{"at": "TRUE", "isAt": "FALSE"}
{"at": "PROBABLE", "isAt": "TRUE"}
"""

print(f"🚀 Running 4-Shot Prompting Experiment via Groq ({MODEL_NAME})...")
for doc in tqdm(cleaned_dev_docs, desc="Processing"):
    doc_copy = json.loads(json.dumps(doc))
    pairs_list_str = "\n".join([f"{i}. Person: {p['pers_mentions_list'][0]} | Place: {p['loc_mentions_list'][0]}" for i, p in enumerate(doc_copy['sampled_pairs'])])

    prompt_content = f"""You are an expert computational historian specializing in relation extraction.
TASK: Determine historical relations ('at' and 'isAt') between the designated Persons and Places based on the text.

CRITICAL LOGICAL CONSTRAINTS:
1. 'at' represents a permanent, structural, institutional, professional, or residency-based geographic connection over time.
2. 'isAt' represents literal immediate physical presence at that place within the narrative moment.
3. If 'at' is FALSE, then 'isAt' MUST be FALSE.
4. Use 'PROBABLE' for 'at' when strong contextual, regional, or family/professional affiliation implies geographic connectivity without explicit absolute proof.

---
HERE ARE REFERENCE EXAMPLES TO EMULATE SYSTEM REASONING:
{examples_4_shot}
---

TARGET TEXT FOR ANALYSIS:
"{doc_copy['text']}"

Evaluate the relationships for the following requested pairs exactly in sequence. Return JSON objects containing keys "at" and "isAt".

PAIRS TO EVALUATE:
{pairs_list_str}"""

    # Hit the API instead of relying on the local broken environment
    chat_completion = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt_content}],
        model=MODEL_NAME,
        temperature=0.0
    )

    raw_response = chat_completion.choices[0].message.content
    batch_preds = harvest_json_robust(raw_response)

    for i, pair in enumerate(doc_copy['sampled_pairs']):
        raw_pred = batch_preds[i] if i < len(batch_preds) else {"at": "FALSE", "isAt": "FALSE"}
        if isinstance(raw_pred, list) and len(raw_pred) > 0: raw_pred = raw_pred[0]
        if not isinstance(raw_pred, dict): raw_pred = {"at": "FALSE", "isAt": "FALSE"}

        at_val = str(raw_pred.get("at", "FALSE")).upper()
        isAt_val = str(raw_pred.get("isAt", "FALSE")).upper()

        pair['at'] = "TRUE" if "TRUE" in at_val else ("PROBABLE" if "PROBABLE" in at_val else "FALSE")
        pair['isAt'] = "TRUE" if "TRUE" in isAt_val else "FALSE"
        if pair['at'] == "FALSE": pair['isAt'] = "FALSE"

    results.append(doc_copy)

# Save the predictions back to drive
output_filename = "llama_8b_4_SHOT_cleaned_results.jsonl"
full_output_path = os.path.join(DRIVE_REPO_PATH, output_filename)
with open(full_output_path, "w", encoding="utf-8") as f:
    for r in results: f.write(json.dumps(r) + "\n")

print(f"\n✅ Experiment complete! Output saved to: {full_output_path}")

# Run evaluation script directly
print("\n📊 EVALUATION SCORER METRICS:")
!cd {DRIVE_REPO_PATH} && python3 scripts/file_scorer_evaluation.py \
    --gold_data_file data/sandbox/en-dev.jsonl \
    --predictions_file {output_filename}
````

````python
import json
import os
import time
from openai import OpenAI
from google.colab import userdata

# 1. INITIALIZATION & CONFIGURATION
# Set your OpenRouter key and paths
OPENROUTER_API_KEY = userdata.get("OPENROUTER_API_KEY")  # Paste your key inside quotes
LLAMA_MODEL_NAME = "meta-llama/llama-3.3-70b-instruct"

# Explicit paths matching your drive layout
input_file_llama = '/content/drive/MyDrive/colab_data/HIPE-2026-data/data/sandbox/en-dev-cleaned.jsonl'
output_file_llama_few_shot = '/content/drive/MyDrive/colab_data/HIPE-2026-data/data/sandbox/llama_70b_standard_4_shot_results.jsonl'

# Initialize OpenRouter Client
client_openrouter = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=OPENROUTER_API_KEY,
)

FEW_SHOT_EXAMPLES = """
### OFFICIAL DEFINITIONS ###
- AT: TRUE (lives there, from there, or long-term role). PROBABLE (a plausible inference can be made from context, such as being in a nearby city or associated with a local port). FALSE (no evidence).
- ISAT: TRUE (physically there during the story). FALSE (not there or event happened in the past).

### REASONING EXAMPLES ###

Example 1: Plausible Geographical Inference
Text: "The British admiral, Purvis, with about 12 ships of war, was at Cadiz."
Pair: Person: Purvis | Place: Cadiz Bay
Reasoning: The text states Purvis is at the city of Cadiz. Cadiz is the primary city of Cadiz Bay. Therefore, it is highly plausible to infer a relation to the Bay.
Output: {"at": "PROBABLE", "isAt": "FALSE"}

Example 2: Contextual Association (Port/City)
Text: "A letter has been received by Mr. Weightman, of Fell’s Point, from capt. Johnson, of this port."
Pair: Person: captain Johnson | Place: Fell’s Point
Reasoning: Capt. Johnson is "of this port." Fell's Point is the specific port neighborhood mentioned in the same sentence. It is plausible to infer he is associated with Fell's Point.
Output: {"at": "PROBABLE", "isAt": "FALSE"}

Example 3: Explicit Presence
Text: "JOE called himself a negro man, who is the property of Mr. Pain Wareing, and was committed to the jail of Alexandria."
Pair: Person: JOE | Place: Alexandria
Reasoning: Joe is physically in the jail located in Alexandria right now.
Output: {"at": "TRUE", "isAt": "TRUE"}

Example 4: Total Lack of Evidence
Text: "Captain Taylor arrived from Cadiz and mentioned that King Joseph had taken Seville."
Pair: Person: King Joseph | Place: Philadelphia
Reasoning: King Joseph is in Spain (Seville/Cadiz); he has no relation to Philadelphia where the ship arrived.
Output: {"at": "FALSE", "isAt": "FALSE"}
"""

def get_batch_predictions_llama_openrouter_few_shot(text, pairs):
    """
    Sends all pairs for a document to Llama 3.3 (70B) via OpenRouter, optimized for speed and JSON,
    using a few-shot prompt.
    """
    pairs_list_str = "\n".join([f"{i}. Person: {p['pers_mentions_list'][0]} | Place: {p['loc_mentions_list'][0]}" for i, p in enumerate(pairs)])

    prompt = f"""You are an expert historian.
    TASK: Determine relations between Persons and Places based strictly on the text.

    CRITICAL RULE: Use 'PROBABLE' whenever a plausible inference can be made.
    If a person is in a city, they are 'PROBABLE' for the surrounding Bay or County.
    If a person is 'of this port', they are 'PROBABLE' for the port name mentioned nearby.

    {FEW_SHOT_EXAMPLES}

    Analyze this text: "{text}"
    Return a JSON object with the key "results". Inside, provide an array of objects for each pair in sequence matching the schema: {{"at": "TRUE/PROBABLE/FALSE", "isAt": "TRUE/FALSE"}}.

    PAIRS:
    {pairs_list_str}"""

    try:
        response = client_openrouter.chat.completions.create(
            messages=[{"role": "user", "content": prompt}],
            model=LLAMA_MODEL_NAME,
            temperature=0.0,
            top_p=1.0,
            max_tokens=1024,
            response_format={"type": "json_object"}
        )

        content = response.choices[0].message.content

        if content is None:
            return [{"at": "FALSE", "isAt": "FALSE"}] * len(pairs)

        # Clear markdown formatting if present
        if "```json" in content:
            content = content.split("```json")[1].split("```")[0].strip()
        elif "```" in content:
            content = content.split("```")[1].split("```")[0].strip()

        parsed = json.loads(content)
        if isinstance(parsed, dict) and "results" in parsed and isinstance(parsed["results"], list):
            return parsed["results"]
        elif isinstance(parsed, list):
            return parsed
        elif isinstance(parsed, dict):
            for key in parsed:
                if isinstance(parsed[key], list):
                    return parsed[key]
            return [parsed]
        else:
            print(f"  ⚠️ Unexpected JSON type: {type(parsed)}. Returning default.")
            return [{"at": "FALSE", "isAt": "FALSE"}] * len(pairs)

    except Exception as e:
        print(f"  ⚠️ Error in OpenRouter batch (Llama 3.3 70B): {e}")
        return [{"at": "FALSE", "isAt": "FALSE"}] * len(pairs)


# 2. EXECUTION LOOP
print(f"🚀 Starting {LLAMA_MODEL_NAME} (OpenRouter) 4-Shot Batch Inference...")
start_time_llama_few_shot = time.time()

processed_docs_llama_few_shot = []
os.makedirs(os.path.dirname(output_file_llama_few_shot), exist_ok=True)

with open(input_file_llama, "r") as f_in:
    lines = f_in.readlines()
    for line_idx, line in enumerate(lines):
        doc = json.loads(line)
        doc_id = doc.get("document_id", "unknown")
        pairs = doc.get('sampled_pairs', [])

        print(f"🔄 [{line_idx+1}/{len(lines)}] Processing Doc: {doc_id} ({len(pairs)} pairs)...", end="")
        loop_start = time.time()

        if pairs:
            batch_results_llama_few_shot = get_batch_predictions_llama_openrouter_few_shot(doc['text'], pairs)

            for i, pair in enumerate(pairs):
                pred = batch_results_llama_few_shot[i] if i < len(batch_results_llama_few_shot) else {"at": "FALSE", "isAt": "FALSE"}

                # Coerce values to clean upper-case strings matching schema expectations
                pair['at'] = str(pred.get("at", "FALSE")).upper()
                pair['isAt'] = str(pred.get("isAt", "FALSE")).upper()

                # Re-apply structural rule just in case model broke coherence
                if pair['at'] == "FALSE":
                    pair['isAt'] = "FALSE"

                pair['at_explanation'] = "Llama-3.3-70B-OpenRouter 4-Shot"
                pair['isAt_explanation'] = "Llama-3.3-70B-OpenRouter 4-Shot"

        processed_docs_llama_few_shot.append(doc)
        print(f" Done! ({round(time.time() - loop_start, 2)}s)")
        time.sleep(2)  # OpenRouter pacing safety margin

# 3. SAVE RESULTS
with open(output_file_llama_few_shot, "w") as f_out:
    for doc in processed_docs_llama_few_shot:
        f_out.write(json.dumps(doc) + "\n")

print(f"\n✅ SUCCESS! Total time: {round(time.time() - start_time_llama_few_shot, 2)} seconds.")
print(f"Final file saved to: {output_file_llama_few_shot}")
````

````python
import json
import os
import time
import torch
from openai import OpenAI
from sentence_transformers import SentenceTransformer
from google.colab import userdata

# =====================================================================
# 1. INITIALIZATION
# =====================================================================
OPENROUTER_API_KEY = userdata.get("OPENROUTER_API_KEY")
LLAMA_MODEL_NAME = "meta-llama/llama-3.3-70b-instruct"
drive_repo_path = '/content/drive/MyDrive/colab_data/HIPE-2026-data'
input_file_en = f'{drive_repo_path}/data/sandbox/en-dev-cleaned.jsonl'
output_file_en_dynamic = f'{drive_repo_path}/data/sandbox/llama_70b_en_dynamic_4shot.jsonl'

client_openrouter = OpenAI(base_url="https://openrouter.ai/api/v1", api_key=OPENROUTER_API_KEY)
embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# =====================================================================
# 2. DEFINING THE 10-SHOT EXTREME EXEMPLAR POOL
# =====================================================================
TRAINING_POOL_EN = [
    {"text": "The national society of Belgian railways held its first general meeting in Brussels yesterday.", "person": "Railway Minister Lippens", "place": "Belgium", "at": "TRUE", "isAt": "TRUE"},
    {"text": "In the afternoon, Mr. Müller was seen in the market square in Leipzig.", "person": "Mr. Müller", "place": "Leipzig", "at": "TRUE", "isAt": "TRUE"},
    {"text": "The envoy from Austria arrived late at the Paris station yesterday evening.", "person": "envoy", "place": "Austria", "at": "TRUE", "isAt": "FALSE"},
    {"text": "Kaiser Wilhelm traveled to Berlin, while General Bonaparte gathered troops near Paris.", "person": "General Bonaparte", "place": "Berlin", "at": "FALSE", "isAt": "FALSE"},
    {"text": "The French ambassador gave a formal reception at the embassy yesterday evening.", "person": "ambassador", "place": "France", "at": "TRUE", "isAt": "FALSE"},
    {"text": "The inhabitants of Zurich protested violently against the new customs regulations.", "person": "inhabitants", "place": "Zurich", "at": "TRUE", "isAt": "TRUE"},
    {"text": "The Crown Prince returned to Vienna after his long journey through Italy.", "person": "Crown Prince", "place": "Italy", "at": "FALSE", "isAt": "FALSE"},
    {"text": "While Dr. Weber remained in Munich, Professor Schmidt traveled directly onward to London.", "person": "Dr. Weber", "place": "London", "at": "FALSE", "isAt": "FALSE"},
    {"text": "The management of the Krupp works in Essen issued a statement regarding the strikes.", "person": "management", "place": "Essen", "at": "TRUE", "isAt": "TRUE"},
    {"text": "In a letter from Hamburg, Mr. Schulze reports on the devastating storm surge.", "person": "Mr. Schulze", "place": "Hamburg", "at": "TRUE", "isAt": "TRUE"}
]
pool_texts = [item["text"] for item in TRAINING_POOL_EN]
pool_embeddings = embedding_model.encode(pool_texts, convert_to_tensor=True)

# =====================================================================
# 3. PROMPT GENERATOR
# =====================================================================
def build_isolated_prompt(query_text, pairs_list_str, target_task):
    query_embedding = embedding_model.encode(query_text, convert_to_tensor=True)
    similarities = embedding_model.similarity(query_embedding, pool_embeddings)[0]
    top_4_indices = torch.topk(similarities, k=4).indices.cpu().numpy()
    dynamic_examples_str = ""
    for rank, idx in enumerate(top_4_indices, 1):
        example = TRAINING_POOL_EN[idx]
        val = example['at'] if target_task == "at" else example['isAt']
        dynamic_examples_str += f"\nExample {rank}:\nText: \"{example['text']}\"\nPair: Person: {example['person']} | Place: {example['place']}\nOutput: {{\"results\": [{{\"{target_task}\": \"{val}\"}}]}}"

    task_def = ("- AT: TRUE/PROBABLE/FALSE" if target_task == "at" else "- ISAT: TRUE/FALSE")
    schema = f'{{"{target_task}": "..."}}'
    return f"Task: {task_def}\n\nExamples:\n{dynamic_examples_str}\n\nText: \"{query_text}\"\n\nPairs:\n{pairs_list_str}\n\nReturn JSON: {{\"results\": [{schema}, ...]}}"

# =====================================================================
# 4. API PIPELINE
# =====================================================================
def get_isolated_predictions(text, pairs, target_task):
    pairs_list_str = "\n".join([f"{i}. Person: {p['pers_mentions_list'][0]} | Place: {p['loc_mentions_list'][0]}" for i, p in enumerate(pairs)])
    prompt = build_isolated_prompt(text, pairs_list_str, target_task)
    fallback = [{target_task: "FALSE"}] * len(pairs)
    try:
        response = client_openrouter.chat.completions.create(
            messages=[{"role": "user", "content": prompt}],
            model=LLAMA_MODEL_NAME, temperature=0.0, response_format={"type": "json_object"}
        )
        raw = response.choices[0].message.content.strip()
        if "```" in raw:
            raw = raw.split("```")[1].replace("json", "").strip()
        return json.loads(raw).get("results", fallback)
    except:
        return fallback

# =====================================================================
# 5. EXECUTION ENGINE
# =====================================================================
raw_docs = []
with open(input_file_en, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip(): raw_docs.append(json.loads(line))

print("Track 1/2: Processing AT...")
at_results = {}
for doc in raw_docs:
    at_results[doc['document_id']] = get_isolated_predictions(doc['text'], doc.get('sampled_pairs', []), "at")

print("Track 2/2: Processing ISAT...")
isat_results = {}
for doc in raw_docs:
    isat_results[doc['document_id']] = get_isolated_predictions(doc['text'], doc.get('sampled_pairs', []), "isAt")

# =====================================================================
# 6. COMBINATION & EXPORT
# =====================================================================
final_docs = []
for doc in raw_docs:
    d_id = doc['document_id']
    pairs = doc.get('sampled_pairs', [])
    for i, pair in enumerate(pairs):
        pair['at'] = at_results[d_id][i].get('at', 'FALSE') if i < len(at_results[d_id]) else 'FALSE'
        pair['isAt'] = isat_results[d_id][i].get('isAt', 'FALSE') if i < len(isat_results[d_id]) else 'FALSE'
    final_docs.append(doc)

with open(output_file_en_dynamic, "w") as f:
    for d in final_docs: f.write(json.dumps(d) + "\n")

print("\n📊 RUNNING EVALUATION:")
os.system(f"cd {drive_repo_path} && python3 scripts/file_scorer_evaluation.py --gold_data_file data/sandbox/en-dev-cleaned.jsonl --predictions_file data/sandbox/llama_70b_en_dynamic_4shot.jsonl")
````
